```python
# CUSTOMER PURCHASE + PRODUCT RECOMMENDATION SYSTEM
# ============================================================
#
# PART 1  -> Online Shoppers Intention
#           Predict whether customer will purchase
#
# PART 2  -> Online Retail
#           Recommend similar products
#
# FINAL   -> Both systems are stored inside ONE model file
#
# OUTPUT:
# customer_purchase_recommendation.pkl
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

from model.customer_syestem import CustomerSystem
import pickle

from sklearn.model_selection import train_test_split

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.preprocessing import (
    OneHotEncoder,
    MinMaxScaler
)

from sklearn.feature_selection import (
    SelectKBest,
    chi2
)

from sklearn.linear_model import LogisticRegression

from sklearn.tree import DecisionTreeClassifier

from sklearn.ensemble import (
    RandomForestClassifier,
    GradientBoostingClassifier
)

from xgboost import XGBClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

from sklearn.metrics.pairwise import cosine_similarity


# ============================================================
# 2. LOAD ONLINE SHOPPERS DATASET
# ============================================================

dataset = pd.read_csv(
    "https://archive.ics.uci.edu/ml/machine-learning-databases/00468/online_shoppers_intention.csv"
)

print("Online Shoppers Dataset:")
print(dataset.shape)

print(dataset.head())


# ============================================================
# 3. CREATE X AND Y
# ============================================================

x = dataset.iloc[:, :-1]

y = dataset["Revenue"]


print("\nFeatures:")
print(x.columns)

print("\nTarget:")
print(y.name)


# ============================================================
# 4. TRAIN TEST SPLIT
# ============================================================

train_x, test_x, train_y, test_y = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)


# ============================================================
# 5. PREPROCESSING
# ============================================================

tf1 = ColumnTransformer(

    transformers=[

        (
            "categorical",

            OneHotEncoder(
                sparse_output=False,
                handle_unknown="ignore"
            ),

            [10, 15]
        )

    ],

    remainder="passthrough"
)


# ============================================================
# 6. SCALING
# ============================================================

tf2 = MinMaxScaler()


# ============================================================
# 7. FEATURE SELECTION
# ============================================================

tf3 = SelectKBest(
    score_func=chi2,
    k=15
)


# ============================================================
# 8. CREATE MODELS
# ============================================================

models = {

    "Logistic Regression":
        LogisticRegression(
            max_iter=1000
        ),

    "Decision Tree":
        DecisionTreeClassifier(
            random_state=42
        ),

    "Random Forest":
        RandomForestClassifier(
            random_state=42
        ),

    "Gradient Boosting":
        GradientBoostingClassifier(
            random_state=42
        ),

    "XGBoost":
        XGBClassifier(
            random_state=42,
            eval_metric="logloss"
        )
}


# ============================================================
# 9. TRAIN ALL MODELS
# ============================================================

pipes = {}

results = []


for name, model in models.items():

    print("\n===================================")
    print("Training:", name)
    print("===================================")

    pipe = Pipeline([

        ("preprocessing", tf1),

        ("scaling", tf2),

        ("feature_selection", tf3),

        ("model", model)

    ])


    pipe.fit(
        train_x,
        train_y
    )


    prediction = pipe.predict(
        test_x
    )


    accuracy = accuracy_score(
        test_y,
        prediction
    )

    precision = precision_score(
        test_y,
        prediction,
        zero_division=0
    )

    recall = recall_score(
        test_y,
        prediction,
        zero_division=0
    )

    f1 = f1_score(
        test_y,
        prediction,
        zero_division=0
    )


    pipes[name] = pipe


    results.append({

        "Model": name,

        "Accuracy": accuracy,

        "Precision": precision,

        "Recall": recall,

        "F1 Score": f1

    })


# ============================================================
# 10. MODEL RESULTS
# ============================================================

results_df = pd.DataFrame(
    results
)


results_df = results_df.sort_values(
    by="F1 Score",
    ascending=False
).reset_index(
    drop=True
)


print("\n\n===================================")
print("MODEL PERFORMANCE")
print("===================================")

print(
    results_df.to_string(
        index=False
    )
)


# ============================================================
# 11. SELECT BEST MODEL
# ============================================================

best_model_name = results_df.iloc[0]["Model"]

best_purchase_pipeline = pipes[
    best_model_name
]


print("\n===================================")
print("BEST MODEL")
print("===================================")

print(
    "Best Model:",
    best_model_name
)


print(
    "Best F1 Score:",
    results_df.iloc[0]["F1 Score"]
)


# ============================================================
# 12. LOAD ONLINE RETAIL DATASET
# ============================================================

print("\n\n===================================")
print("LOADING RETAIL DATASET")
print("===================================")


retail = pd.read_excel(
    "Online Retail.xlsx"
)


print(
    "Original Retail Shape:",
    retail.shape
)


# ============================================================
# 13. RETAIL DATA CLEANING
# ============================================================

retail = retail.dropna(
    subset=["CustomerID"]
)


retail = retail[
    retail["Quantity"] > 0
]


retail = retail[
    retail["UnitPrice"] > 0
]


retail = retail[
    ~retail["InvoiceNo"]
    .astype(str)
    .str.startswith("C")
]


retail = retail.dropna(
    subset=[
        "StockCode",
        "Description"
    ]
)


print(
    "Clean Retail Shape:",
    retail.shape
)


# ============================================================
# 14. CREATE CUSTOMER-PRODUCT MATRIX
# ============================================================

customer_product = retail.pivot_table(

    index="CustomerID",

    columns="StockCode",

    values="Quantity",

    aggfunc="sum",

    fill_value=0

)


print(
    "\nCustomer Product Matrix:",
    customer_product.shape
)


# ============================================================
# 15. TRANSPOSE MATRIX
# ============================================================

product_customer = customer_product.T


print(
    "Product Customer Matrix:",
    product_customer.shape
)


# ============================================================
# 16. CALCULATE COSINE SIMILARITY
# ============================================================

similarity = cosine_similarity(
    product_customer
)


# ============================================================
# 17. CREATE SIMILARITY DATAFRAME
# ============================================================

similarity_df = pd.DataFrame(

    similarity,

    index=product_customer.index,

    columns=product_customer.index

)


print(
    "Similarity Matrix:",
    similarity_df.shape
)


# ============================================================
# 18. CREATE PRODUCT TABLE
# ============================================================

products = retail[
    [
        "StockCode",
        "Description"
    ]
].drop_duplicates(
    subset=["StockCode"]
)


print(
    "\nProducts:",
    products.shape
)


# ============================================================
# 19. RECOMMENDATION FUNCTION
# ============================================================

def recommend_products(
    stock_code,
    n=10
):

    if stock_code not in similarity_df.index:

        print(
            "Product not found."
        )

        return pd.DataFrame()


    similar_products = similarity_df[
        stock_code
    ].sort_values(
        ascending=False
    )


    similar_products = similar_products[
        similar_products.index != stock_code
    ]


    similar_products = similar_products.head(
        n
    )


    recommendations = (
        similar_products
        .reset_index()
    )


    recommendations.columns = [
        "StockCode",
        "Similarity"
    ]


    recommendations = recommendations.merge(

        products,

        on="StockCode",

        how="left"

    )


    recommendations = recommendations[
        [
            "StockCode",
            "Description",
            "Similarity"
        ]
    ]


    return recommendations


# ============================================================
# 20. TEST RECOMMENDATION SYSTEM
# ============================================================

test_product = retail[
    "StockCode"
].iloc[0]


print("\n===================================")
print("RECOMMENDATION TEST")
print("===================================")

print(
    "Selected Product:",
    test_product
)


recommendations = recommend_products(
    test_product,
    n=10
)


print(
    recommendations.to_string(
        index=False
    )
)


# ============================================================
# 22. CREATE MODEL METRICS DICTIONARY
# ============================================================

model_metrics = {}


for row in results:

    model_metrics[
        row["Model"]
    ] = {

        "accuracy": float(
            row["Accuracy"]
        ),

        "precision": float(
            row["Precision"]
        ),

        "recall": float(
            row["Recall"]
        ),

        "f1_score": float(
            row["F1 Score"]
        )

    }


# ============================================================
# 23. CREATE FINAL CUSTOMER SYSTEM
# ============================================================
# NOTE: passing the FULL pipes dict (all 5 trained models), not
# just the best one, so the API can serve predictions from any
# model the user selects, not only the best-performing one.

system = CustomerSystem(

    purchase_pipelines=
        pipes,

    best_model=
        best_model_name,

    similarity_df=
        similarity_df,

    products=
        products,

    model_metrics=
        model_metrics

)


# ============================================================
# 24. TEST PURCHASE PREDICTION
# ============================================================

sample_customer = test_x.iloc[
    [0]
]


purchase_prediction = (
    system.predict_purchase(
        sample_customer
    )[0]
)


purchase_probability = (
    system.purchase_probability(
        sample_customer
    )
)


confidence = (
    system.confidence(
        sample_customer
    )
)


print("\n===================================")
print("PURCHASE PREDICTION TEST")
print("===================================")


print(
    "Best Model:",
    system.best_model
)


print(
    "Prediction:",
    bool(purchase_prediction)
)


print(
    "Purchase Probability:",
    purchase_probability
)


print(
    "Confidence:",
    confidence
)


# ============================================================
# 25. SAVE ONE COMPLETE MODEL
# ============================================================

with open(
    "customer_purchase_recommendation.pkl",
    "wb"
) as f:

    pickle.dump(
        system,
        f
    )


print("\n===================================")
print("MODEL SAVED")
print("===================================")

print(
    "customer_purchase_recommendation.pkl"
)


# ============================================================
# 26. CHECK SAVED MODEL
# ============================================================

print("\nTesting saved model...")


with open(
    "customer_purchase_recommendation.pkl",
    "rb"
) as f:

    loaded_system = pickle.load(
        f
    )


print(
    "Loaded object type:",
    type(loaded_system)
)


print(
    "Best Model:",
    loaded_system.best_model
)


print(
    "Available Models:",
    list(loaded_system.purchase_pipelines.keys())
)


# ============================================================
# 27. TEST RECOMMENDATION FROM SAVED MODEL
# ============================================================

saved_recommendations = (
    loaded_system.recommend(
        test_product,
        n=10
    )
)


print("\n===================================")
print("SAVED MODEL RECOMMENDATIONS")
print("===================================")


print(
    saved_recommendations.to_string(
        index=False
    )
)


# ============================================================
# 28. FINAL SYSTEM INFORMATION
# ============================================================

print("\n===================================")
print("FINAL SYSTEM")
print("===================================")

print(
    "Purchase Model:",
    loaded_system.best_model
)

print(
    "Recommendation System: Active"
)

print(
    "Model Metrics:"
)

for model_name, metrics in (
    loaded_system.model_metrics.items()
):

    print(
        "\n",
        model_name
    )

    print(
        "Accuracy:",
        round(
            metrics["accuracy"],
            4
        )
    )

    print(
        "Precision:",
        round(
            metrics["precision"],
            4
        )
    )

    print(
        "Recall:",
        round(
            metrics["recall"],
            4
        )
    )

    print(
        "F1 Score:",
        round(
            metrics["f1_score"],
            4
        )
    )


print("\n===================================")
print("DONE")
print("===================================")
```

```python
import pickle
import __main__


class CustomerSystem:

    def __init__(
        self,
        purchase_pipeline,
        similarity_df,
        products
    ):
        self.purchase_pipeline = purchase_pipeline
        self.similarity_df = similarity_df
        self.products = products

    def predict_purchase(self, X):
        return self.purchase_pipeline.predict(X)

    def recommend(self, stock_code, n=10):

        similarity = self.similarity_df[stock_code]

        similar = similarity.sort_values(
            ascending=False
        )

        similar = similar[
            similar.index != stock_code
        ].head(n)

        return self.products[
            self.products["StockCode"].isin(
                similar.index
            )
        ]


__main__.CustomerSystem = CustomerSystem


with open(
    "customer_purchase_recommendation.pkl",
    "rb"
) as f:

    system = pickle.load(f)


print("SYSTEM TYPE:")
print(type(system))

print("\nPURCHASE PIPELINE TYPE:")
print(type(system.purchase_pipeline))

print("\nPURCHASE PIPELINE:")
print(system.purchase_pipeline)

print("\nSIMILARITY TYPE:")
print(type(system.similarity_df))

print("\nPRODUCTS TYPE:")
print(type(system.products))

print("\nPRODUCTS SHAPE:")
print(system.products.shape)
```
